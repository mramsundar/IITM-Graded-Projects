```python
import os
import sys
sys.path.append(os.path.dirname(os.getcwd()))
from query_helper import get_connection, execute_script, import_csv

def start():
    root_dir = os.path.dirname(os.getcwd())
    raw_data_dir = os.path.join(root_dir, "Raw data")

    conn = get_connection()

    if conn:
        execute_script(conn, "create_tables.sql")
        import_csv(conn, os.path.join(raw_data_dir, "patients.csv"), "patients")
        import_csv(conn, os.path.join(raw_data_dir, "visits.csv"), "visits")
        import_csv(conn, os.path.join(raw_data_dir, "billing.csv"), "billing")
        conn.close()

start()
```
